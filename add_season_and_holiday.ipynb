```python
import os
import glob
import pandas as pd

INPUT_DIR = "final_data"
OUTPUT_DIR = "final_data_updated"

os.makedirs(OUTPUT_DIR, exist_ok=True)

def get_season(month_period_str):
    # Split the "YYYY-MM" by the hyphen and returns the month number
    month = int(str(month_period_str).split("-")[1])

    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Fall"


parquet_files = glob.glob(os.path.join(INPUT_DIR, "*.parquet"))
print(f"Found {len(parquet_files)} parquet files.")

for file_path in parquet_files:
    df = pd.read_parquet(file_path)

    if df.empty:
        continue

    # Add the season column using get_season
    df['season'] = df['month_period'].apply(get_season)

    # Add the holiday column (setting all rows to "no" for now)
    df['holiday'] = "no"

    filename = os.path.basename(file_path)
    out_path = os.path.join(OUTPUT_DIR, filename)
    df.to_parquet(out_path, index=False)

print(f"Done! All updated files have been saved to the '{OUTPUT_DIR}' directory.")
```
